## 1. Introduction

The travel behaviour of the Dutch population has changed over time. Different age groups have different travel needs and preferences, which may affect their choice of transport mode and travel purpose. In addition, several policy interventions have been introduced that have specifically affected certain age groups. These interventions may have influenced how people travel.

Understanding whether such interventions are associated with changes in travel behaviour is relevant for policy makers, because transport policy often targets specific age groups, such as older people and students.

This report uses CBS data for 1999–2025 to analyse the change in transport mode choice of the whole population and three age-specific interventions: the medical driving licence check, the state pension age and the student travel product. By comparing an affected age group with a similar, unaffected group before and after each intervention, we look for patterns that may be related to these interventions.

## 2. Research Objective

The objective of this research is to examine how travel purposes and transport mode choice in the Netherlands have changed between 1999 and 2025, with a particular focus on age-specific policy interventions. First, overall changes in transport mode choice across the Dutch population are analysed. Next, the research examines changes in car-driving behaviour among people aged 70 and older following the change in driving licence renewal requirements in 2014, this intervention changed the minimum age of medical check-up for a drivers license renewal from 70 to 75 years of age. It also investigates whether the gradual increase in the state pension age is associated with changes in work-related trips among people aged 65–70. Finally, the research examines whether extending the student travel product to MBO students under 18 is associated with changes in public transport use among young people.

By comparing travel behaviour across age groups and over time, this research aims to identify patterns and associations that may be related to these policy interventions. 

This leads to the following main research question and sub-questions:

**Main research question**

To what extent have age-specific interventions influenced travel purposes and transport mode choice in the Netherlands between 1999 and 2025?

**Sub-questions**

1. How has the choice of travel mode changed between 1999-2025 among the entire population in the Netherlands?
2. How did car driving among people aged 70+ change around the increase in the medical examination age in 2014?
3. Is the gradual increase of the state pension age (AOW) from 65 to 67 years (2013–2024) associated with an increase in work trips among people aged 65–70, compared with people aged 60–65?
4. Is the extension of the free student travel product to MBO students under 18 (1 January 2017) associated with an increase in train and bus/tram/metro use among 12–18 year olds, compared with 18–25 year olds?

### 6.4 SQ4: Free Student Travel below 18 years
To evaluate the impact of extending the free student travel product on January 1 2017, the combined public transport use (train, bus, tram and metro) of the agegroup 12 to 18 years was compared against a control agegroup of 18 to 25 years. To compare the effect of the policy change, a baseline period is selected between 2014 and 2016.  This three-year average provides a stable reference point right before the policy change. Using this period avoids incidental annual fluctuations in mobility, while taking in account the socio-economic growth and changes over the years. In the analysis, the years 2020 and 2021 were intentionally excluded because of the disruption to regular travel behaviour, as result of the COVID-19 pandemic. 

Figure 9 shows the absolute public transport trips per person per day, figure 10 shows the relative change indexed to 2016 as baseline. The vertical dashed line marks the 2017 policy change, highlighting the differences in travel behaviour between the agegroups.

The analyzed mobility data reveals two notable patterns that formulate an answer to the subquestion:

* **Short-term Effect** [2017-2019]: In the three years immediately after the policy change, the number of PT trips among the 12-18 agegroup increased by 4.1% compared to 2016. In the same period, the number of PT trips of the control group decreased by 1.0%. This difference could suggest a positive correlation between the policy change and the PT use of the target group.
* **Post-COVID Effect** [2022-2025]: After the pandemic, total trips in public transport of both agegroups have not fully recovered. But there is a significant difference between the both agegroups. Compared to the baseline period (2014-2016), the 18-25 agegroup travels has decreased 21.5%, where the decrease of the 12-18 agegroup is only 14.7%.

These observations suggest that the extension of student travel product to MBO students under 18 is associated with an increase in trips per day among the 12-18 agegroup, and that this group remained more attached to PT after the pandemic than the comparison group.

However, this association does not prove that the policy caused the change. The main limitation within this data analysis is the level of detail of the demographic groups. The 12-18 agegroup is significantly broader than just the MBO student group. A large share of this category consists of secondary school students who did not receive the benefit on the policy change. Therefore, it is difficult to isolate the specific effect of the intervention within this broader demographic group. Finally, other developments, such as the rise of the e-bike, may have influenced public transport use in both age groups.

## 7. Conclusion

This research examined to what extent age-specific interventions influenced travel purposes and transport mode choice in the Netherlands between 1999 and 2025. Using Dataset 1, the overall development of transport mode choice was analysed first, after which three interventions aimed at specific age groups were studied by comparing travel behaviour before and after each intervention.

Between 1999 and 2025, transport mode choice in the Netherlands gradually changed. Car use, both as a driver and as a passenger, declined over the period, while bicycle use remained relatively stable and walking increased, particularly after 2020. The largest short-term change occurred during the COVID-19 pandemic, when the number of trips decreased for most transport modes. Train and bus/tram/metro use recovered only partly afterwards and had not returned to pre-pandemic levels by 2025. 

For people aged 70 and older, the number of car-driver trips per person per day was about 11% higher in 2014–2019 than in 2008–2013. However, this increase had already started well before the medical driving licence check age was raised from 70 to 75 in 2014. The annual growth did not accelerate after the change, and the value in 2019 was almost equal to the extended trend of 2004–2013. Therefore, the data show no clear evidence that the change in the driving licence renewal age affected car driving among people aged 70+.

The gradual increase of the state pension age does not appear to have had a distinct effect on commuting among people aged 65–70. Although their number of commute trips rose by about 35% after 2013, a comparable rise of 39% was found among the 60–65 agegroup, who were not directly affected by policy, and this growth had already begun before 2013. The higher number of commute trips among older people therefore seems to reflect a general tendency to remain in the workforce for longer, rather than a response to the policy alone, although the policy may have contributed to this development.

The extension of the free student travel product to MBO students under 18 in 2017 is associated with an increase in public transport use among the 12–18 agegroup. In 2017–2019, their train and bus/tram/metro trips increased by about 4%, while those of the 18–25 comparison group decreased by 1%. After the pandemic, public transport use declined in both groups, but less strongly among the 12–18 agegroup, 14.7%, than among the 18–25 agegroup, 21.5%. Although this pattern is consistent with a positive effect of the policy, MBO students form only a small part of the 12–18 group, so the effect of the intervention itself cannot be isolated.

Taken together, the influence of age-specific interventions on travel behaviour in the Netherlands appears to be limited. Only the student travel product shows a difference between the affected group and its comparison group that is consistent with an effect of the intervention. For the driving licence renewal age and the state pension age, the observed increases are in line with trends that were already ongoing or that also occurred in the comparison group. Overall, long-term developments, such as older people driving and working for longer, and major events such as the COVID-19 pandemic seem to have had a larger impact on travel behaviour than the individual policy interventions. Because the data are smoothed, cover broad age groups and do not allow causal conclusions, these findings should be interpreted as associations rather than proven effects, as further explained in the discussion.

## 8. Discussion

The conclusions of this research indicate that the evaluated age-specific policy interventions have a limited identifiable impact on overall transport mode choices, compared to broader societal trends and the COVID-19 pandemic. However, these findings should be interpreted by acknowledging the methodological choices, the precision of the data used, and the complexity of the subject.

A primary limitation affecting the reliability of the results is the use of a descriptive before-and-after comparison. Travel behaviour is ultimately human behaviour, which is always multi-causal and related to a wider range of external influences beyond the government interventions. While the research did identify associations (like the different post-COVID recovery trends of the 12-18 and 18-25 agegroups), these observations are not binding for the policy evaluation. The analysis cannot isolate the policy effect from other parallel macroeconomic or societal developments.

Furthermore, the precision of the research is limited by the decision to use macro-level secondary data. This approach was selected to keep the research feasible within the available time and technical capabilities. Therefore, there is a structural mismatch between the broad age categories, provided by CBS and the specific demographic targets of the policies. For example, the medical driving licence check specifically targeted people reaching the age of 70, but the data only allowed for an analysis of the entire 70+ agegroup. Furthermore, the 12-18 agegroup includes a majority of secondary school students who were completely unaffected by the MBO student travel product. However, this level of detail fits the research objective. Identifying shifts within these broad age categories lays the basis for future research to examine specific demographic agegroups more closely.

Another notable methodological simplification that affects the precision of the modal split analysis is the merging of regular and e-bikes into a single category. The rapid development and widespread adoption of e-bikes have fundamentally changed modern travel behaviour. E-bikes lower the physical effort required to cycle and also provides full flexibility in travel choices, compared to public transport. By treating all bicycles as a single mode for transport, the analysis may miss underlying shifts in behaviour that act independently of the policy intervention. 

Finally, the temporal scope of the research introduced uncertainties regarding long-term structural changes. To assess the lasting effects of the interventions, the analysis compared a pre-COVID baseline (2014-2016) with a post-COVID period (2022-2025). While travel patterns are likely to have been settled into a new standard following the pandemic, the available data for this period only spans a few years. Although comparing these figures to pre-COVID baselines enabled the analysis, the short post-pandemic observation period makes it difficult to confirm whether this new standard is a permanent structural shift.
