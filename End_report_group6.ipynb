# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

The Dutch population is ageing quickly: the share of people aged 65 and older keeps growing, and older people travel for different purposes and with different modes than younger people. At the same time, several policy interventions have changed the conditions under which specific age groups travel, such as the gradual increase of the state pension age (AOW) and the extension of the free student travel product to MBO students under 18. In addition, the COVID-19 pandemic strongly disrupted travel behaviour, and not every age group was affected in the same way.

The objective of this research is to quantify how population ageing and age-specific interventions are associated with changes in travel purposes and transport mode choice in the Netherlands between 1999 and 2025. First, we separate the change in national travel patterns into a part caused by the changing age structure of the population (*ageing effect*) and a part caused by changed travel behaviour within age groups (*behaviour effect*). Second, we examine whether the travel behaviour of older people has changed over time. Third, we analyse three events that affected specific age groups, by comparing the affected age group with a similar, unaffected comparison group before and after the event.

This leads to the following main and sub-questions:

**Main research question**

How are population ageing and age-specific interventions associated with changes in travel purposes and transport mode choice in the Netherlands between 1999 and 2025?

**Sub-questions**

*Population ageing*
1. To what extent are changes in national travel purposes and transport mode choice explained by the changing age structure of the population (ageing effect), and to what extent by changed travel behaviour within age groups (behaviour effect)?
2. How has the travel behaviour of older people (65+) changed between 1999 and 2025 in terms of number of trips, trip distance, travel purposes and transport modes?
3. Is the gradual increase of the state pension age (AOW) from 65 to 67 years (2013–2024) associated with an increase in work trips among people aged 65–70, compared with people aged 60–65?
4. How did the COVID-19 pandemic (2020–2021) affect the travel behaviour of different age groups, and did older people recover to pre-pandemic levels as fast as younger people?
5. Is the extension of the free student travel product to MBO students under 18 (1 January 2017) associated with an increase in train and bus/tram/metro use among 12–18 year olds, compared with 18–25 year olds?

# Data Pipeline

To research the impact of the changing fuel prices on modal split, two datasets from CBS will be compared. Firstly, the fuel prices will be obtained and be generalized into average price per month. Then, the mobility dataset will be used to obtain the number of trips per mode per person per month. The number of trips per person per mode per month will be compared to the fuel price per month to analyze the impact. After that, the trips will be categorized into mandatory trips and flexible trips and the comparison will be repeated, to see what type of trip is affected the most.

## 4. Basic pre-conditions

To keep the analysis clear and reproducible, the following pre-conditions and assumptions apply to all datasets used in this study.

### 4.1 General pre-conditions
- **Geographical scale:** the Netherlands as a whole (national averages). Regional differences (provinces, urban vs. rural) are out of scope.
- **Temporal scale:** January 2018 – December 2024, analysed at a monthly resolution. This gives 84 monthly observations per transport mode.
- **Provisional data:** CBS marks the 2024 figures as provisional (`2024*`). They are used as they are, and this is mentioned as a limitation.
- **Population:** CBS mobility figures are for residents of the Netherlands aged 6 years and older.
- **Survey-based data:** the mobility figures come from the ODiN travel survey (a sample, not a full count). We use the point estimate (`Waarde`) and leave out the confidence margins.
- **COVID-19 period:** defined as March 2020 – February 2022 (from the first lockdown until most measures were lifted). In this period travel behaviour was mostly driven by government measures, not by prices. The analysis is therefore done twice: once for the full period and once without the COVID period (sub-question 3).
- **Other influences are not modelled:** factors such as public transport fares, weather, working from home, e-bike growth and economic conditions also affect mode choice. They are not part of the model, so the results show correlations, not proven causal effects.

### 4.2 Fuel prices
- **Source:** CBS *Pompprijzen motorbrandstoffen; brandstofsoort, per dag* (80416NED). The daily prices are converted into a*monthly average price so that they match the monthly mobility data.
- **National average price:** pump prices differ per location (e.g. motorway vs. non-motorway stations). We use the national average price.
- **Fuel types:** only Euro95 (petrol) and diesel. LPG is out of scope because only a small share of passenger cars use it.
- **Combined fuel price:** one fuel price per month is calculated as a weighted average of Euro95 and diesel:
  `P_fuel = w_petrol × P_Euro95 + w_diesel × P_diesel`
  The weights are the shares of petrol and diesel cars in the Dutch passenger car fleet (CBS vehicle fleet statistics). These weights are kept constant over 2018–2024.
- **Pump price as paid by the traveller:** prices include taxes and VAT, so changes such as the temporary fuel excise reduction from 2022 are already included. Prices are nominal (not corrected for inflation). An inflation-corrected price (CPI) can be added as a robustness check.
- **Electric vehicles:** electric car users are not directly affected by pump prices. Their share of the fleet was still small in this period, so this effect is not modelled separately.
- **Timing:** we assume travellers respond to the fuel price of the same month. A delayed reaction (1–3 months lag) can be tested as an extra analysis.

### 4.3 Transport modes
- **Modes in scope:** car (driver), train, bus/tram/metro, bicycle (including e-bike) and walking.
- **Car = car driver:** the car driver is the one who pays for fuel, so the effect of fuel prices is analysed on *car (driver)*. Car passenger can be shown for context.
- **Out of scope:** other modes (ferry, moped, etc. – CBS category *Overige vervoerwijze*).
- **Main mode only:** CBS assigns each trip one main mode (*hoofdvervoerwijze*), the mode used for the longest distance of the trip. Trips with more than one mode (e.g. bike + train) count only under their main mode.
- **Modal split definition:** the modal split per month is the share of each mode in the total number of trips (trips per person per day for a mode ÷ total trips per person per day). The share based on distance (passenger-kilometres) is used as a check.

### 4.4 Trip purposes
The CBS travel purposes (*reismotieven*) are grouped into three categories:

| Category | CBS travel purposes |
|---|---|
| **1. Mandatory trips** | Van en naar het werk (commute), Zakelijk/beroepsmatig (business), Onderwijs volgen/cursus (education) |
| **2. Shopping** | Winkelen, boodschappen doen |
| **3. Flexible trips** | Visite/logeren (visiting), Uitgaan/sport/hobby (leisure, sports), Toeren/wandelen (touring), Diensten/persoonlijke verzorging (services), Overige reismotieven (other) |

- Shopping is kept as a separate group because it is partly mandatory (groceries) and partly flexible.
- **Data limitation:** CBS publishes trip purposes per mode only as yearly values (table 84710NED), not per month. The analysis per trip purpose is therefore done on a yearly level (7 data points), while the main fuel-price analysis is done on a monthly level.
