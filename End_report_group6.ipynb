## 1. Introduction

The travel behaviour of the Dutch population has changed over time, influenced by demographic developments and changes in government policies. Different age groups have different travel needs and preferences, which may affect their choice of transport mode and travel purpose. In addition, several policy interventions have specifically affected certain age groups. These interventions may have influenced how people travel.

Understanding whether such interventions are associated with changes in travel behaviour is relevant for policy makers, because transport policy often targets specific age groups, such as older people and students.

This report uses CBS data for 1999–2025 to analyse the change in transport mode choice of the whole population and three age-specific interventions: the medical driving licence check, the state pension age and the student travel product. By comparing an affected age group with a similar, unaffected group before and after each intervention, we look for patterns that may be related to these interventions.

## 2. Research Objective

The objective of this research is to examine how travel purposes and transport mode choice in the Netherlands have changed between 1999 and 2025, with a particular focus on age-specific policy interventions. First, overall changes in transport mode choice across the Dutch population are analysed. Next, the research examines changes in car-driving behaviour among people aged 70 and older following the change in driving licence renewal requirements in 2014, this intervention changed the minimum age of medical check-up for a drivers license renewal from 70 to 75 years of age. It also investigates whether the gradual increase in the state pension age is associated with changes in work-related trips among people aged 65–70. Finally, the research examines whether extending the student travel product to MBO students under 18 is associated with changes in public transport use among young people.

By comparing travel behaviour across age groups and over time, this research aims to identify patterns and associations that may be related to these policy interventions. 

This leads to the following main research question and sub-questions:

**Main research question**

To what extent have age-specific interventions influenced travel purposes and transport mode choice in the Netherlands between 1999 and 2025?

**Sub-questions**

1. How has the choice of travel mode changed between 1999-2025 among the entire population in the Netherlands?
2. To what extend has car driving behaviour changed among the 70+ group after 2014 as a result of the increase in the driving license renewal age?
3. Is the gradual increase of the state pension age (AOW) from 65 to 67 years (2013–2024) associated with an increase in work trips among people aged 65–70, compared with people aged 60–65?
4. Is the extension of the free student travel product to MBO students under 18 (1 January 2017) associated with an increase in train and bus/tram/metro use among 12–18 year olds, compared with 18–25 year olds?

### 4.2 Travel data (84755NED)
- The data comes from four successive CBS surveys: OVG (1999-2003), MON (2004-2009), OViN (2010-2017) and ODiN (2018-2025). CBS corrected the breaks between the surveys with a time-series model, so the full period can be compared. As a side-effect, sudden changes are partly smoothed, so effects of interventions may look more gradual than they were.
- The data covers residents of the Netherlands aged 6 years and older who live in private households. People in care homes and other institutions are not included, which mainly affects the oldest age group.
- Only trips within the Netherlands are included. Trips abroad and holiday trips are excluded.
- The main measure is the number of trips per person per day. A trip is defined as, a movement with usage of one single transport mode. A bike ride to the station followed by a train ride counts as two trips.

### 4.3 Transport modes and travel purposes
- Modes in scope: car (driver), car (passenger), train, bus/tram/metro, bicycle and walking. Other modes are part of the total but are not analysed separately.
- The bicycle includes the electric bicycle, because 84755NED does not split them.
- The modal split is the share of each mode in the total number of trips per person per day.
- The travel purposes of 84755NED are used as they are: work (commuting), education, shopping, leisure and other.

### 4.4 Age groups and population
- Age groups follow 84755NED: 6-12, 12-18, 18-25, 25-30, 30-40, 40-50, 50-60, 60-65, 65-70 and 70+.
- 70+ is one group, so people aged 75+ cannot be analysed separately.
- No population data is used, so the results describe the average trips per person in each age group.

### 6.1 SQ1: Overall Travel Behaviour 1999-2025
In Figure X, the absolute differences in the use of different modes of transport for the entire population of the Netherlands can be seen. Several things stand out in this figure.

First, the biggest changes can be seen around 2020. The average number of trips decreases for several transport modes during this period. This could be explained by the COVID-19 pandemic, when people travelled less because they worked from home more often and there were restrictions on social activities. For example, the number of trips by car as a driver decreases from approximately 0.95 trips per person per day in 2019 to around 0.80 in 2020.

Another interesting observation is the increase in walking. Around 2009, walking reaches a temporary peak of approximately 0.68 trips per person per day. Walking also increases around 2020 and remains at a relatively high level in the following years. This could be related to changes in people's daily routines, although the graph alone cannot confirm the exact reason for this increase.

Car use shows a gradual decline over the period. Both car drivers and car passengers make fewer trips per person per day towards the end of the period than at the beginning. Bicycle use, on the other hand, remains relatively stable, despite some fluctuations between years. This suggests that the average number of bicycle trips has not changed as much as car use over the period.

Train, bus, tram and metro use also show a decline around 2020. Although there is some recovery afterwards, the number of trips does not immediately return to the levels seen before the pandemic. The category other modes of transport also changes over time. This category includes transport modes that are not shown separately in the graph, but more information would be needed to determine what causes these changes.

Overall, the graph shows that transport behaviour in the Netherlands has changed between 2000 and 2025. The most noticeable changes occur around 2020, while the longer-term trends show a decline in car use and relatively stable bicycle use. However, these are absolute values, so they do not yet show whether a transport mode has become more or less important compared with all other trips. ### 6.4 SQ4: Free Student Travel below 18 years
To evaluate the impact of extending the free student travel product on January 1 2017, the combined public transport use (train, bus, tram and metro) of the agegroup 12 to 18 years was compared against a control agegroup of 18 to 25 years. To compare the effect of the policy change, a baseline period is selected between 2014 and 2016.  This three-year average provides a stable reference point right bofere the policy change. Using this period avoids incidental annual fluctuations in mobility, while taking in account the socio-economic growth and changes over the years. In the analysis, the years 2020 and 2021 were intentionally excluded because of the disruption to regular travel behaviour, as result of the COVID-19 pandemic. 

In the Figure below, the top graph shows the absolute public transport trips per person per day, while the bottom graph shows the relative change indexed to 2016 as baseline. The vertical dashed line markes the 2017 policy change, highlighting the differences in travel behaviour between the agegroups.

![Absolute changes and index in PT use](Output\SQ4-Complete.png)

The analyzed mobility data reveals two notable patterns that formulate an answer to the subquestion:

* **Short-term Effect** [2017-2019]: In the three years immediately after the policy change, the number of PT trips among the 12-18 agegroup increased by 4.1% compared to 2016. In the same period, the number of PT trips of the control group decreased by 1.0%. This difference could suggest a positive correlation between the policy change and the PT use of the target group.
* **Post-COVID Effect** [2022-2025]: After the pandemic, total trips in public transport of both agegroups have not fully recovered. But there is a significant difference between the both agegroups. Compared to the baseline period (2014-2016), the 18-25 agegroup travels has decreased 21.5%, where the decrease of the 12-18 agegroup is only 14.7%.

Based on these observations, it can be concluded that the extension of student travel product to MBO students under 18 not only resulted in an increase in trips per day, but also retained this demographic group in PT more effectively after the pandemic than the control group.

However, this conclusion must be interpreted with caution, as this association does not prove a direct correlation. The main limitation within this data analysis is the level of detail of the demographic groups. The 12-18 agegroup is significantly broader than just the MBO student group. A large share of this category consits of secondary school students who do not recieve the benefit on the policy change. Therefore, it is difficult to isolate the specific effect of the intervention within this broader demographic group. Finally, other factors influencing the travel behaviour of the agegroups, such as the rise of e-bikes, may also have influece the long-term decline of PT trips.
