# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

The primary objective of this research is to model and analyze the relationship between pump fuel prices and modal shift in the Netherlands. By linking fuel price data to national passenger travel patterns, this study aims to assess whether rising fuel costs drive travelers toward alternative modes. Furthermore, the research investigates the travel demand across different trip purposes, distinguishing between mandatory and flexible trips to identify which types of trips show the largest change in travel choices. 

This leads to the following main and sub questions:

**Main research Question**

How do changes in fuel prices impact the modal split, and how does this effect differ across various trip purposes?

**Subquestions**
1. Which alternative transport modes see the largest increase in use when pump prices rise?
2. To what extent does the shift in mode choice differ between mandatory trips and flexible trips?
3. What is the difference in data when excluding the Covid-19 pandemic time frame.
4. What is the difference between daily and monthly travel data trip?

# Data Pipeline

To research the impact of the changing fuel prices on modal split, two datasets from CBS will be compared. Firstly, the fuel prices will be obtained and be generalized into average price per month. Then, the mobility dataset will be used to obtain the number of trips per mode per person per month. The number of trips per person per mode per month will be compared to the fuel price per month to analyze the impact. After that, the trips will be categorized into mandatory trips and flexible trips and the comparison will be repeated, to see what type of trip is affected the most.
