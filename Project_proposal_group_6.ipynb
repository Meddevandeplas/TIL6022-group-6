# Scale
### Geographical Scale 
For this research we will be looking on national level. So we will be looking at the Netherlands.

### Temporal Scale
The research will cover the period from 2018 to 2024. This seven-year period provides sufficient data to analyse developments and identify potential relationships over time. It is important to note that the COVID-19 pandemic occurred within this period and may have significantly influenced both oil prices and transport prices. Therefore, the potential impact of the pandemic will be taken into account when processing and analysing the data.
