# Research Topic

Researching the effect of the Russia - Ukraine war on the fly routes between Europe an Asia 

# Research Objective

### Main Research Question
How has the closure of Ukrainian and Russian airspace folling the 2022 invasion of Ukraine affected flight distances between Europe and Asia?

### Sub Questions
- Which flight routes between Europe and Asia have experienced the largest increase in flight distance due to the airspace closures?
- How has the increase in flight distance affected CO₂ emissions on these routes?
- What are the additional costs for airlines resulting from the increased flight distances?
